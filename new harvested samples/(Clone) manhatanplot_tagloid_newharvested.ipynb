# Manhattan plots + taglo ID annotations (module-based workflow)

This notebook is a **thin runner** that calls functions from `manhattan_module` to generate **Manhattan plots** from GWAS summary statistics and to annotate/highlight **taglo IDs** (taglotype identifiers) for significant signals.

The idea is simple: keep the notebook light (configuration + function calls) and keep the plotting / GWAS logic inside the reusable module.

## What this notebook does
- Loads the project config path and output directory
- Gets the list of traits directly from the **GWAS table** (via `manhattan_module.get_traits`)
- For each trait:
  - loads a lightweight GWAS subset (significant SNPs + sampled background)
  - includes `taglo_id` columns as needed
  - calls `manhattan_module` plotting functions to create:
    
    - taglo-highlighted Manhattan plots 
- Saves plot PNGs to the configured output folder


All of that logic lives in **`manhattan_module.py`**.

## Inputs
This notebook relies on:
- `manhattan_module.py`
- A config YAML file (`config_mixed.yaml`) containing:
  - GWAS table path/name
  - other project paths if needed


The GWAS table is expected to contain at least:
- `trait` (or `phenotype`)
- `chrom`
- `start` (or position column mapped in the module call)
- `p_wald`
- 


## Outputs
 
  - lead-taglo / taglo-highlight Manhattan plots
