# Manhattan plots with effect direction (beta): GWAS signal strength + sign

This notebook generates **Manhattan plots** from GWAS summary statistics, but with an extra focus on the **effect size (beta)**: not only where the strongest associations are, but also whether the allele’s effect is **positive or negative**.

## What this notebook does
- **Load GWAS summary stats** (e.g., GEMMA/LMM outputs) containing at least:
  - `trait` (VOC/phenotype name)
  - `chrom` (chromosome)
  - `start` or `position` (base‑pair coordinate)
  - `p_wald` (p-value)
  - `beta` (effect size / direction)
- **Clean and standardize columns** (casting chrom/position, dropping missing p-values or betas).
- **Compute −log10(p)** for plotting signal strength.
- **Create beta-aware Manhattan plots**:
  - y-axis = −log10(p)
  - x-axis = genome position (often concatenated by chromosome)
  - points can be **colored by beta sign** (positive vs negative) or by beta magnitude (depending on the code in the notebook)
  -  significance line (e.g., p = 1e-6)
- **Export figures** (PNG) to the configured output directory, usually one plot per trait (or per selected trait subset).

## Why this is useful
A standard Manhattan plot shows *where* the signal is. Adding beta lets us see 
*how the allele shifts the trait* (increase vs decrease), which is handy when comparing loci across traits or when selecting candidates for dosage-effect validation.

## Inputs (expected)
- GWAS table/dataframe with `trait`, `chrom`, `start/position`, `p_wald`, and `beta`.


## Outputs
- Beta-aware Manhattan plot PNGs saved to the chosen output directory.
