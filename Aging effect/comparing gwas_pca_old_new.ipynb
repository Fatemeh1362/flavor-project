# Shared loci

```python
%sql
SELECT
    chrom,
    FLOOR(CAST(start_old AS BIGINT) / 100000) * 100000 AS locus_start,
    MIN(p_old) AS best_p_old,
    MIN(p_new) AS best_p_new,
    COUNT(*) AS snp_count
FROM (
    SELECT 
        a.chrom,
        a.start AS start_old,
        b.start AS start_new,
        a.p_wald AS p_old,
        b.p_wald AS p_new
    FROM `bmqg`.`gwas`.`run_local_20260313_old` a
    JOIN `bmqg`.`gwas`.`run_local_20260314_new` b
      ON a.chrom = b.chrom
     AND ABS(CAST(a.start AS BIGINT) - CAST(b.start AS BIGINT)) <= 50000
    WHERE a.p_wald < 5e-4
      AND b.p_wald < 5e-4
) t
GROUP BY chrom, FLOOR(CAST(start_old AS BIGINT) / 100000)
ORDER BY best_p_old
LIMIT 20;
```

# storage-specific locus

```python
%sql
SELECT 
    a.chrom,
    a.start AS start_old,
    a.p_wald AS p_old
FROM `bmqg`.`gwas`.`run_local_20260313_old` a
WHERE a.p_wald < 5e-6
AND NOT EXISTS (
    SELECT 1
    FROM `bmqg`.`gwas`.`run_local_20260314_new` b
    WHERE a.chrom = b.chrom
    AND ABS(CAST(a.start AS BIGINT) - CAST(b.start AS BIGINT)) <= 50000
    AND b.p_wald < 5e-6
)
ORDER BY a.p_wald
LIMIT 50;
```
