# Gwas for pca_new potato

This notebook performs a genome-wide association study (GWAS) using the GEMMA_taglotype pipeline.

```python

import sys
import pandas as pd
sys.path.append("/Workspace/Shared_Repos")
from gwas_analysis.gemma_taglotype import GEMMA_taglotype
import yaml
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import shutil, os
```

```python

CONFIG_PATH = "/Volumes/bmqg/default_bronze/fatemeh/config_mixed.yaml"

with open(CONFIG_PATH, "r") as f:
    CONFIG = yaml.safe_load(f)

PHENO_PATH = "/Volumes/bmqg/default_bronze/fatemeh/final_project/pca_traits_for_gwas_new.csv"

df_pheno = pd.read_csv(PHENO_PATH)

varieties = df_pheno["Variety"]

X = df_pheno.drop(columns=["Variety"])

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

Xs = StandardScaler().fit_transform(X)

pca = PCA(n_components=5)
scores = pca.fit_transform(Xs)

pca_traits = pd.DataFrame(
    scores,
    columns=["PC1","PC2","PC3","PC4","PC5"]
)

pca_traits.insert(0, "Variety", varieties)
pca_traits.set_index("Variety", inplace=True)
```

```python

shutil.rmtree("/tmp/gemma_local", ignore_errors=True)

local_tmp = "/tmp/gemma_local"
os.makedirs(local_tmp, exist_ok=True)

print(local_tmp)
```

```python
run_path = "/tmp/tmp_gemma/run_20260314_new"

os.makedirs(run_path, exist_ok=True)
print("Run path ready:", run_path)
```

```python
gemma = GEMMA_taglotype(
    pca_traits,
    min_allele_freq=0.007,
    max_allele_freq=0.8,
    max_bad_fraction=0.1,
    max_combinations=4,
    ploidy=4,
    run_id="run_local_20260314_new"
)

gemma.run_gwas()
```
