# Manhaan plot
This notebook provides an interactive dashboard for exploring Manhattan plots generated for different traits. It loads precomputed plot images from a specified directory and maps them to their corresponding traits. Using a dropdown widget, users can select a trait and dynamically display its associated Manhattan plot. This allows for quick visual inspection and comparison of genetic association signals across traits without recomputing the plots.

```python


import ipywidgets as widgets
from IPython.display import display, clear_output
import os
import re
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
from PIL import Image

```

```python

IMG_DIR ="/Volumes/bmqg/default_bronze/fatemeh/final_project/main_manhattan_plot_newharvested/"

def safe_name(s):
    return re.sub(r"[^\w\-.]+", "_", str(s)).strip("_")

# map: trait → image path
trait_to_image = {}

for fn in os.listdir(IMG_DIR):
    if fn.endswith(".png"):
     
        trait = fn.split("_", 1)[1].replace(".png", "")
        trait_to_image[trait] = os.path.join(IMG_DIR, fn)

print("Images found:", len(trait_to_image))
```

```python
trait_dropdown = widgets.Dropdown(
    options=sorted(trait_to_image.keys()),
    description="Trait:",
    layout=widgets.Layout(width="60%")
)

out = widgets.Output()

def on_trait_change(change):
    with out:
        clear_output(wait=True)

        trait = change["new"]
        img_path = trait_to_image.get(trait)

        print(f"Trait: {trait}")
        print(f"Image: {os.path.basename(img_path)}")

        if not os.path.exists(img_path):
            print(" Image not found")
            return

        img = Image.open(img_path)

        plt.figure(figsize=(16, 5))
        plt.imshow(img)
        plt.axis("off")
        plt.show()

trait_dropdown.observe(on_trait_change, names="value")

display(trait_dropdown, out)
```
