```python
import pandas as pd

CSV_PATH ="/Volumes/bmqg/default_bronze/fatemeh/final_project/csv_outputs/allelic_effect_linearit.csv"

df_all = pd.read_csv(CSV_PATH)

```

```python
print(df_all.columns.tolist())
```

```python
traits = sorted(df_all["trait"].dropna().astype(str).unique().tolist())

print("Number of traits:", len(traits))
print("First 50 traits:")
print(traits[:50])

```

```python
def run_dashboard(trait):
    clear_output(wait=True)

    df = df_all[df_all["trait"] == trait].copy()

    if df.empty:
        print("No data for this trait")
        return

    print(f"Trait: {trait}")
    print(f"Number of loci: {len(df)}")

    display(
        df[[
            "chrom",
            "start_min",
            "start_max",
            "n_taglos",
            "dosages_used",
            "dosage_shape",
            "spearman_rho",
            "linear_r2",
            "delta_extreme",
            "taglo_ids"        
        ]]
        .sort_values("linear_r2", ascending=False)
    )
```

```python

widgets.interact(
    run_dashboard,
    trait=traits
);
```
