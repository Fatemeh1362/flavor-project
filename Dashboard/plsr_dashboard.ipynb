# PLSR dashboard
This notebook implements an interactive dashboard for visualizing Partial Least Squares Regression (PLSR) results across multiple traits. It loads pre-generated PNG plots from a specified workspace directory and links each plot to its corresponding trait. Using a dropdown widget, users can select a trait and instantly display its associated visualization. This setup enables efficient exploration and comparison of PLSR outputs without rerunning analyses.

```python
import matplotlib
matplotlib.rcdefaults()
import os
import ipywidgets as widgets
from IPython.display import display, clear_output, Image
```

```python
# ============================================================
# GLOBAL-QTL PLSR DASHBOARD (PNG-ONLY, FROM WORKSPACE)
# ============================================================
import os
import ipywidgets as widgets
from IPython.display import display, Image, clear_output

# correct path for Unity Catalog Volume
IMG_BASE = "/Volumes/bmqg/default_bronze/fatemeh/final_project/plsr/"

# collect PNG files
png_files = sorted([
    f for f in os.listdir(IMG_BASE)
    if f.lower().endswith(".png")
])

if not png_files:
    raise RuntimeError("No PNG files found in IMG_BASE")

traits = [os.path.splitext(f)[0] for f in png_files]

print(f"Found {len(traits)} traits")

trait_dd = widgets.Dropdown(
    options=traits,
    description="Trait:",
    layout=widgets.Layout(width="420px")
)

out = widgets.Output()

def show_png(trait):

    with out:
        clear_output(wait=True)

        img_path = os.path.join(IMG_BASE, f"{trait}.png")

        print(f"Trait: {trait}")
        print("-" * 40)

        if os.path.exists(img_path):
            display(Image(filename=img_path))
        else:
            print("PNG not found")
            print(img_path)

trait_dd.observe(lambda c: show_png(c["new"]), names="value")

show_png(trait_dd.value)

display(widgets.VBox([trait_dd, out]))
```
